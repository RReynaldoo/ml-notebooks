```python
#Libraries for the assignment

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import preprocessing
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
#Reading the wine csv file
df = pd.read_csv('wine.csv')
```

```python
#Defining the target and input variables
y = df['Wine']
X = df[['Phenols', 'Flavanoids', 'Nonflavanoid.phenols']]
```

```python
#displaying them in array format
X_arr = X.to_numpy()
y_arr = y.to_numpy()
print("Input variables array (first 5 rows):\n", X_arr[:5])
print("\nTarget variable array (first 5 values):\n", y_arr[:5])
```

```python
#splititng the data into tersting and training set
X_train, X_test, y_train, y_test = train_test_split(X_arr, y_arr, test_size=0.25, random_state=42)
```

```python
#printing them
print("\nTraining set inputs (first 5 rows):\n", X_train[:5])
print("\nTesting set inputs (first 5 rows):\n", X_test[:5])
```

```python
#Scaling the data
scaler = preprocessing.StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#implementing the svm classifier
clf = SVC(kernel='linear', C=1.0)
clf.fit(X_train_scaled, y_train)
```

```python
#using the plot to show the classifier
plt.figure(figsize=(8, 6))
plt.scatter(X_train_scaled[:, 0], X_train_scaled[:, 1], c=y_train, cmap='viridis', edgecolors='k')
plt.title('SVM Classification (Phenols vs Flavanoids)')
plt.xlabel('Scaled Phenols')
plt.ylabel('Scaled Flavanoids')
plt.show()
```

```python
#lastly implementing and displaying hte matric and accuracy of scores
y_pred = clf.predict(X_test_scaled)
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nAccuracy Score:", accuracy_score(y_test, y_pred))
```
